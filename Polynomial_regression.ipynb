```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#generate same synthetic dp or reproducitiblity of result

np.random.seed(1)
```

```python
#generate synthetic x

x = 2*np.random.rand(100,1)
```

```python
x
```

```python
# generating random syntetic y using x
y = 4 + 3*x + 1.5*x **2 + np.random.randn(100,1)
```

```python
#train_test_split
```

```python
from sklearn.model_selection import train_test_split
train_test_split = train_test_split(x,y,test_size=0.2,random_state=42)
x_train,x_test,y_train,y_test = train_test_split
```

```python
x_test.shape,y_test.shape,x_train.shape,y_train.shape
```

```python
# perform polynomial regression
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly_features = PolynomialFeatures(degree=2,include_bias=False)
x_train_poly = poly_features.fit_transform(x_train)
x_test_poly = poly_features.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
lin_reg = LinearRegression()
lin_reg.fit(x_train_poly,y_train)
```

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(y_test,lin_reg.predict(x_test_poly))
```

```python
plt.scatter(x,y)
plt.show()
```

```python
x_range = np.linspace(0, 2, 100).reshape(-1, 1)
X_range_poly = poly_features.transform(x_range)

degree = 2  # Defined based on PolynomialFeatures(degree=2)
plt.plot(x_range, lin_reg.predict(X_range_poly), color='red', label=f'Polynomial Regression (Degree {degree})')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Polynomial Regression')
plt.legend()
plt.show()
```
